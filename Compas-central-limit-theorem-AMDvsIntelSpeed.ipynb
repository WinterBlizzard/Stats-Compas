## Apply Central Limit Theorem

```python
# storage of means
means_data = {"AMD": [], "Intel": []}

np.random.seed(42)

# sampling means
for _ in range(50): 
    for c in ["AMD", "Intel"]:
        sample = np.random.choice(df[c], size=30, replace=True)
        means_data[c].append(np.mean(sample))
```

```python
df_means = pd.DataFrame(means_data)
df_means
```

```python
df_means.to_csv(r"dataset/AMDvsIntel-clt-dataset.csv", index=False)
print("CLT dataset saved!")
```

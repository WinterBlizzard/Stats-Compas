<h1>Histogram</h1>

```python
plt.hist(df['Current'], bins=5,
        alpha=0.8,
        color= 'red',
        edgecolor='white',)

plt.title('Histogram of Current Values',
         fontsize=14, font='Arial', color='#333333')

plt.xlabel('Current (mA)', fontsize=12, color='#333333')
plt.ylabel('Frequency', fontsize=12, color='#333333')
plt.show()
```

<h1>Normal Distribution</h1>

```python
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)
```

```python
x_min = df['Current'].min()
x_max = df['Current'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)

plt.plot(x, p, 'red', linewidth=2)
plt.hist(df['Current'], bins=5,
        density=True,
        alpha=0.8,
        color= 'aqua',
        edgecolor='white',)


plt.title('Normal Distribution of Current Values', fontsize=14, font='Arial', color='#333333')
plt.xlabel('Current (mA)', fontsize=12, color='#333333')
plt.ylabel('Density', fontsize=12, color='#333333')
plt.show()
```

<h1>Standard Normal Distribution</h1>

```python
x_max = df['Z-score'].max()
x_min = df['Z-score'].min()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, df['Z-score'].mean(), df['Z-score'].std(ddof=1))

plt.plot(x, p, 'red', linewidth=2)


plt.title('Normal Distribution of Z-score Values', fontsize=14, font='Arial', color='#333333')
plt.xlabel('Z-score', fontsize=12, color='#333333')
plt.ylabel('Probability  Density', fontsize=12, color='#333333')
plt.show()
```
